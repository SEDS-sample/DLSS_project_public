```python
import pandas as pd 
from functools import reduce
```

```python
df_Lars = pd.read_excel("../data/gold_standard_Lars.xlsx")
df_Leander = pd.read_excel("../data/gold_standard_Leander.xlsx")
df_Leon = pd.read_excel("../data/gold_standard_Leon.xlsx")
df_Philipp = pd.read_excel("../data/gold_standard_Philipp.xlsx")

```

```python
df_Lars = df_Lars.rename(columns={"gold_rating":"isco_Lars"})
df_Leander = df_Leander.rename(columns={"isco":"isco_Leander"})
df_Leon = df_Leon.rename(columns={"isco":"isco_Leon"})
df_Philipp = df_Philipp.rename(columns={"isco":"isco_Philipp"})
```

```python
dfs = [df_Lars,df_Leander,df_Leon,df_Philipp]

dfs_reduced = [d[["job_id", "title", "text_final", d.columns[d.columns.str.startswith("isco_")][0]]] for d in dfs]

merged = reduce(lambda left, right: pd.merge(left, right, on=["job_id", "title", "text_final"], how="outer"), dfs_reduced)
```

```python
merged.to_csv("../data/merged_labels.csv", index=False)
```
