```python
import pandas as pd
from collections import Counter
```

```python
df = pd.read_csv("../data/merged_labels.csv")
df
```

```python
label_cols = ["isco_Lars", "isco_Leander", "isco_Leon", "isco_Philipp"]

def majority_vote(row):
    votes = [str(row[c]) for c in label_cols if pd.notna(row[c])]

    if not votes:
        return "check"
    
    count = Counter(votes)

    top_code, top_n = count.most_common(1)[0]

    if top_n > len(votes) / 2:
        return top_code
    else:
        return "check"

df["isco_final"] = df.apply(majority_vote, axis=1)
df
```

```python
print(df["isco_final"].value_counts().get("check", 0), "Cases to check")
```

```python
df.to_excel("../data/gold_standard_labels_final.xlsx", index=False)
```
